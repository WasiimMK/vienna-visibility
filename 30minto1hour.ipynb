```python
import numpy as np
import netCDF4 as nc
```

```python
from google.colab import drive, files
drive.mount('/content/gdrive')
```

```python
data_list = []

for i in range(2009, 2020) :
  
  filename = '/content/gdrive/MyDrive/Colab Notebooks/ERA30MinVertical/t_vertical/t_'
  filename = filename + str(i) + '.nc'
  
  data = nc.Dataset(filename)
  cc = data['t'][:]

  for x in range(8760) :

    temp_list = []

    for y in range(13) :

      a = cc[x][y][0][0]
      b = cc[x][y][0][1]
      c = cc[x][y][1][0]
      d = cc[x][y][1][1]

      mean = (a + b + c + d)/4
      temp_list.append(mean)

    data_list.append(temp_list)
```

```python
filename = '/content/gdrive/MyDrive/Colab Notebooks/ERA30MinVertical/t_vertical/t_2020.nc'

data = nc.Dataset(filename)
  
cc = data['t'][:]

for x in range(2195) :

  temp_list = []

  for y in range(13) :

    a = cc[x][y][0][0]
    b = cc[x][y][0][1]
    c = cc[x][y][1][0]
    d = cc[x][y][1][1]

    mean = (a + b + c + d)/4
    temp_list.append(mean)

  data_list.append(temp_list)
```
